# Problem Definition
Goal: Build a model to classify whether a tumor is malignant or benign.

```python
#Data Collection (from sklearn)

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
data = load_breast_cancer()
print(data.DESCR)
```

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target  # 1 = Malignant, 0 = Benign
df.sample(5)
```

# Data Preprocessing

```python
print(df.isnull().sum())         # No missing values
print("Duplicates:", df.duplicated().sum())
```

There are no missing or duplicates values in this dataset

# Exploratory Data Analysis (EDA)
Correlation Matrix

```python
corr = df.corr()
corr
```

Here we can see that fews columns are highly co-relatted

```python
df = df.drop(columns=["mean perimeter","mean area","worst perimeter","worst area","worst texture","worst radius"], axis = 1)
df.head()
```

```python
df["target"].value_counts().plot(kind="bar")
plt.title("Target Distribution (0 = Benign, 1 = Malignant)")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

# Feature Engineering

```python
# mean_radius_ratio = mean radius / mean perimeter
df["radius_texture_ratio"] = df["mean radius"] / df["mean texture"]
df.head()
```

# Standardization

```python
X = df.drop(columns=["target"])
y = df["target"]
```

```python
# Histogram Before Standardization

X.iloc[:, :6].hist(figsize=(10,6))
plt.suptitle("Histograms BEFORE Standardization")
plt.tight_layout()
plt.show()
```

```python
# Apply StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
X_scaled.iloc[:, :5].hist(figsize=(10,6))
plt.suptitle("Histograms AFTER Standardization")
plt.tight_layout()
plt.show()
```

```python
# Split the Data Set
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

# Model Selection
 Only logistic regression.

# Model Traning

```python
model = LogisticRegression(max_iter=2000)
model.fit(X_train, y_train)
```

# Model Evaluation

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1-score:", f1_score(y_test, y_pred))
```
